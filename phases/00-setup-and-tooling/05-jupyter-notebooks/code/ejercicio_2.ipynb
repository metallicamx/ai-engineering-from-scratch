# Análisis de Datos: Carga y Visualización

Este cuaderno demuestra el flujo de trabajo completo:
1. Carga de datos desde un archivo CSV.
2. Exploración del DataFrame.
3. Generación de un gráfico estadístico.

> **Nota:** Asegúrate de tener un archivo llamado `datos.csv` en la misma carpeta que este cuaderno.   

```python
import pandas as pd
import matplotlib.pyplot as plt

# Configuración para que los gráficos se muestren dentro del cuaderno
%matplotlib inline

# URL del dataset Iris
datos_url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"

# 1. Cargar el archivo CSV directamente desde la URL
# Eliminamos el bloque try/except local para forzar la carga desde internet
print(f"Cargando datos desde: {datos_url} ...")
df = pd.read_csv(datos_url)

print("Archivo cargado exitosamente desde internet.")

# 2. Mostrar el DataFrame (primeras filas)
df.head()   
```

```python
# 3. Dibujar el gráfico (Adaptado para dataset Iris)
plt.figure(figsize=(10, 6))

# Cambiamos plt.plot por plt.scatter ya que no hay fechas, son datos categóricos/numéricos
plt.scatter(df['petal_length'], df['petal_width'], 
            label='Pétalos', color='blue', alpha=0.7, edgecolors='k')

plt.title('Relación entre Largo y Ancho del Pétalo (Iris Dataset)')
plt.xlabel('Largo del Pétalo (cm)')   # Eje X adaptado
plt.ylabel('Ancho del Pétalo (cm)')   # Eje Y adaptado
plt.grid(True, which='both', linestyle='--', alpha=0.7)
plt.legend()
plt.tight_layout()
plt.show()   
```
