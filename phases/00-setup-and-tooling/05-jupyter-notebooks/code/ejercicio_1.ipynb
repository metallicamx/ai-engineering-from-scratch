## Ejercicio 1

Abre JupyterLab, crea un cuaderno y usa %timeit para comparar la comprensión de listas frente a NumPy al crear un array de 100,000 números aleatorios.

```python
import numpy as np 
```

```python
# 1. Medición de Comprensión de Listas (Python nativo)
# Genera 100,000 números aleatorios usando la librería random estándar dentro de una lista
print("Ejecutando comprensión de listas...")
%timeit [np.random.random() for _ in range(100000)]
```

```python
# 2. Medición de NumPy (Vectorizado)
# Genera 100,000 números aleatorios utilizando la función optimizada de NumPy
print("Ejecutando NumPy...")
%timeit np.random.random(100000)  
```
