# Área de un círculo

La fórmula es: **A = π × r²**

En este notebook calculamos el área para un radio de 5 unidades.

```python
import math

radio = 5
area = math.pi * radio ** 2
print(f"Radio: {radio}")
print(f"Área:  {area:.4f}")
```

## Verificación
